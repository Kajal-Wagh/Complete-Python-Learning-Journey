```python
class Animal:
    def speak(self):
        print("Animal Speaking")

class Dog(Animal):
    def bark(self):
        print("Dog Barking")


# Creating object of Dog class
d = Dog()

# Calling methods
d.speak()   # inherited from Animal
d.bark()    # from Dog class
```

```python
class Person:
    def info(self):
        print("This is a Person")

class Student(Person):
    def study(self):
        print("Student is studying")


# Creating object
s = Student()

s.info()    # inherited method
s.study()   # own method
```

```python
class Vehicle:
    def start(self):
        print("Vehicle is starting")

class Car(Vehicle):
    def drive(self):
        print("Car is driving")


c = Car()

c.start()   # inherited
c.drive()   # own
```

### Abstraction:

```python
# Abstraction: Abstraction is used to hide internal functionality of the users.
The users is only intract with the basic implementation of the function but inner working is hidden

Abstraction Class : The class that consist one or more methods abstract method is called as abstract class
Abstract method : Abstract Method do not contain implementation 
Abstract class can be inherited by the subclass and abstract method gets its defination in the subclass
-The abstraction is useful when we are designing large function.
 -python provides the abc module to used the abstraction in python
ABC- abstract base class

An abstract class contain the both method normal and abstract method 
Abstract class can not be instantiated, we cannot create object of the abstract class
```

```python
from abc import ABC
class classname(ABC):   # Abstract Class
    def abstract_method(self):   # Abstract Method
        pass
```

```python
from abc import ABC
class Car(ABC):
    def milage(self):
        pass
    def show(self):
        print("This is a Car class")

class Tesla(Car):
    def milage(self):
        print("The milage is 30 kmph")
    def show1(self):
        print("This is a Tesla class")

t=Tesla()
t.milage()
t.show1()
t.show()
print()

class Suzuki(Car):
    def milage(self):
        print("The milage is 25 kmph")
    def show1(self):
        print("This is a Suzuki class")

s=Suzuki()
s.milage()
s.show()
s.show1()
```

```python
# plotting a bar chart for gender and its count
ax = sns.countplot(x = 'Gender',data = df,palette='Set1')

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
#Plotting a bar chart for gender vs total amount

sales_gen = df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x = 'Gender',y='Amount',data = sales_gen)
```

```python
# Toatl Amount vs Age group
sales_age = df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.barplot(x='Age Group',y='Amount',data=sales_age,palette='Blues')
```

```python
# total amount/sales from top 10 states

sales_state = df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State',y='Amount',palette = 'viridis')
```
